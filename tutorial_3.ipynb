```python
import numpy as np
arr_2x2 = np.array([[1, 2], [3, 4]])
print("2x2：\n", arr_2x2)
```

```python
random_arr_3x4 = np.random.rand(3, 4)
print("3x4：\n", random_arr_3x4)
```

```python
ones_arr = np.ones((3, 3))
zeros_arr = np.zeros((3, 3))
print("Array of ones：\n", ones_arr)
print("Array of zeros：\n", zeros_arr)
```

```python
random_arr_6x3 = np.random.rand(6, 3)
transposed_arr = random_arr_6x3.T
print("Original 6x3 random array:\n", random_arr_6x3)
print("Transposed array:\n", transposed_arr)
```

```python
n, p, m = 2, 3, 4
arr1 = np.random.rand(n, p)
arr2 = np.random.rand(p, m)
dot_product = np.dot(arr1, arr2)
print("Array 1:\n", arr1)
print("Array 2:\n", arr2)
print("Dot product result:\n", dot_product)
```
